```python
# Loading files
orders = pd.read_csv('orders.csv')
order_details = pd.read_csv('order_details.csv')
pizzas = pd.read_excel('pizzas.xlsx')
pizza_types = pd.read_excel('pizza_types.xlsx') 
```

```python
# Check for column names to connect files
orders.head()
order_details.head()
pizzas.head()
pizza_types.head()
```

```python
# connecting files
df = orders.merge(order_details, on = 'order_id', how = 'left')
df = df.merge(pizzas, on = 'pizza_id', how = 'left')
df = df.merge(pizza_types, on = 'pizza_type_id', how = 'left')
```
